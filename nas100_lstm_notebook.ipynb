# Universal Directory of Open Knowledge Databases: NAS100 LSTM Optimization

This notebook trains and optimizes the Directional LSTM neural network combined with Fair Value Gap (FVG) and 50-EMA trend filtering for NAS100 futures trading.

```python
import os
import json
import numpy as np
import pandas as pd
import torch
from train_nas100_lstm import compute_features, DirectionalLSTM, train_tf_dataset

print('PyTorch CUDA Available:', torch.cuda.is_available())
print('Device Name:', torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU')
```

## Multi-Timeframe Training & Optimization Grid

```python
def optimize_all_timeframes():
    timeframes = ['NQ_in_1_minute.csv', 'NQ_in_5_minute.csv', 'NQ_in_15_minute.csv', 'NQ_in_1_hour.csv']
    results = []
    for tf in timeframes:
        path = os.path.join('NasData', tf)
        if os.path.exists(path):
            df = pd.read_csv(path).rename(columns={'datetime': 'time', 'volume': 'tick_volume'})
            res = train_tf_dataset(df, tf, optimize_grid=True)
            if res:
                results.append({
                    'timeframe': tf,
                    'acc': res['acc'],
                    'brier': res['brier'],
                    'params': res['opt_params']
                })
    return pd.DataFrame(results)

# Run optimization
res_df = optimize_all_timeframes()
print(res_df)
```
